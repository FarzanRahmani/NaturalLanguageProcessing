```python
## Running the spider in Jupyter environment
from scrapy.crawler import CrawlerProcess

class Varzesh3Spider(scrapy.Spider):
    name = "varzesh3"
    start_urls = [
        'https://www.varzesh3.com/'
    ]
    def parse(self, response):
        matches = []
        for match in response.css('div.fixture-result-match'):
            matches.append({
                'date': match.xpath('preceding::div[@class="date-seprator"]/h4/text()').get(),
                'home_team': match.css('div.fixture-result-match-host span::text').get(),
                'away_team': match.css('div.fixture-result-match-guest span::text').get(),
                'score': match.css('div.fixture-result-match-goals::text').get(),
                'match_link': response.urljoin(match.css('a.fixture-result-match-teams::attr(href)').get()),
                'video_link': match.css('div.fixture-result-match-time a::attr(href)').get()
            })
        return matches

## Running the spider in Jupyter Notebook
process = CrawlerProcess({
    'FEEDS': {
        'matches.json': {'format': 'json'},
    },
})
process.crawl(Varzesh3Spider)
process.start()
```

give me a python web crawling code that searches about Iran's histroy and outputs the results in json format.

```python
import requests
from bs4 import BeautifulSoup
import json

def crawl_iran_history():
    # Define the search query and URL
    query = "Iran's history"
    search_url = f"https://duckduckgo.com/html/?q={query}"

    # Send a GET request to the search engine
    response = requests.get(search_url)
    if response.status_code != 200:
        print("Failed to fetch search results")
        return

    # Parse the HTML content
    soup = BeautifulSoup(response.text, 'html.parser')

    # Extract search result titles and links
    results = []
    for result in soup.find_all('a', class_='result__a', limit=10):  # Limit to 10 results
        title = result.text.strip()
        link = result['href']
        results.append({"title": title, "link": link})

    # Output the results in JSON format
    with open("iran_history_results.json", "w", encoding="utf-8") as json_file:
        json.dump(results, json_file, ensure_ascii=False, indent=4)

    print("Results saved to iran_history_results.json")

if __name__ == "__main__":
    crawl_iran_history()
```

```python
import requests
import json
from bs4 import BeautifulSoup
import random
import time
from typing import Dict, List, Any, Optional
import re

class IranHistoryCrawler:
    def __init__(self):
        self.headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
        }
        self.base_urls = [
            "https://en.wikipedia.org/wiki/History_of_Iran",
            "https://fa.wikipedia.org/wiki/تاریخ_ایران",
            "https://www.britannica.com/place/Iran/History"
        ]
        self.results = []
        
    def crawl(self, max_pages: int = 5) -> None:
        """Crawl websites for information about Iran's history."""
        pages_crawled = 0
        
        for base_url in self.base_urls:
            if pages_crawled >= max_pages:
                break
                
            try:
                response = requests.get(base_url, headers=self.headers)
                if response.status_code == 200:
                    soup = BeautifulSoup(response.text, 'html.parser')
                    
                    # Extract links to history-related pages
                    links = self.extract_history_links(soup, base_url)
                    
                    for link in links:
                        if pages_crawled >= max_pages:
                            break
                            
                        try:
                            historical_event = self.extract_event_data(link)
                            if historical_event:
                                self.results.append(historical_event)
                                pages_crawled += 1
                                
                            # Respect crawling etiquette with delay
                            time.sleep(random.uniform(1.0, 2.0))
                            
                        except Exception as e:
                            print(f"Error processing link {link}: {str(e)}")
                
            except Exception as e:
                print(f"Error crawling {base_url}: {str(e)}")
    
    def extract_history_links(self, soup: BeautifulSoup, base_url: str) -> List[str]:
        """Extract links to pages about Iranian history events."""
        links = []
        
        # Find links that might contain historical information
        for a_tag in soup.find_all('a', href=True):
            href = a_tag['href']
            
            # Process relative links
            if href.startswith('/'):
                domain = re.match(r'(https?://[^/]+)', base_url).group(1)
                href = f"{domain}{href}"
            
            # Filter for likely history pages
            history_keywords = ['history', 'battle', 'war', 'dynasty', 'empire', 'ruler',
                               'تاریخ', 'نبرد', 'جنگ', 'سلسله', 'امپراتوری', 'شاه']
            
            if any(keyword in href.lower() or keyword in a_tag.get_text().lower() for keyword in history_keywords):
                links.append(href)
        
        return links[:10]  # Limit to prevent excessive crawling
    
    def extract_event_data(self, url: str) -> Optional[Dict[str, Any]]:
        """Extract structured data about a historical event from a page."""
        try:
            response = requests.get(url, headers=self.headers)
            if response.status_code != 200:
                return None
                
            soup = BeautifulSoup(response.text, 'html.parser')
            
            # Extract title
            title = soup.find('h1')
            if title:
                title = title.get_text().strip()
            else:
                title = url.split('/')[-1].replace('_', ' ')
            
            # Extract description (first paragraph or intro)
            description = ""
            for p in soup.find_all('p'):
                if len(p.get_text().strip()) > 100:  # Get a substantial paragraph
                    description = p.get_text().strip()
                    break
            
            # Try to extract time period
            period = self.extract_period(soup)
            
            # Try to extract location
            location = self.extract_location(soup)
            
            # Extract causes
            causes = self.extract_list_items(soup, ['cause', 'reason', 'علت', 'دلیل'])
            
            # Extract belligerents
            belligerents = self.extract_belligerents(soup)
            
            # Extract result
            result = self.extract_text_with_keywords(soup, ['result', 'outcome', 'نتیجه'])
            
            # Extract casualties
            casualties = self.extract_casualties(soup)
            
            # Extract impact
            impact = self.extract_list_items(soup, ['impact', 'effect', 'consequence', 'تاثیر', 'پیامد'])
            
            # Extract historical significance
            significance = self.extract_text_with_keywords(soup, ['significance', 'important', 'مهم', 'اهمیت'])
            
            # Extract references
            references = self.extract_references(soup)
            
            return {
                "title": title,
                "description": description,
                "period": period,
                "location": location,
                "causes": causes,
                "belligerents": belligerents,
                "result": result,
                "casualties": casualties,
                "impact": impact,
                "historical_significance": significance,
                "references": references,
                "source": {
                    "title": title,
                    "url": url,
                    "publication_date": "N/A"
                }
            }
            
        except Exception as e:
            print(f"Error extracting data from {url}: {str(e)}")
            return None
    
    def extract_period(self, soup: BeautifulSoup) -> Dict[str, str]:
        """Extract time period from the page."""
        period = {"start_year": "نامشخص", "end_year": "نامشخص"}
        
        # Look for date patterns in text
        text = soup.get_text()
        
        # Look for Persian/Arabic year patterns
        persian_years = re.findall(r'(\d{3,4})\s*(هجری|ه\.ق|ه\.ش)', text)
        if persian_years:
            period["start_year"] = persian_years[0][0]
            if len(persian_years) > 1:
                period["end_year"] = persian_years[-1][0]
            else:
                period["end_year"] = period["start_year"]
        
        # Look for Gregorian year patterns
        years = re.findall(r'\b(1[0-9]{3}|20[0-2][0-9])\b', text)
        if years and not persian_years:
            period["start_year"] = years[0]
            if len(years) > 1:
                period["end_year"] = years[-1]
            else:
                period["end_year"] = period["start_year"]
                
        return period
    
    def extract_location(self, soup: BeautifulSoup) -> Dict[str, Any]:
        """Extract location information."""
        location = {
            "province": "نامشخص",
            "city": "نامشخص",
            "coordinates": {
                "latitude": None,
                "longitude": None
            }
        }
        
        # Look for common Iran provinces and cities in the text
        text = soup.get_text()
        
        provinces = [
            "تهران", "اصفهان", "فارس", "خراسان", "آذربایجان", "کرمان", "خوزستان", 
            "گیلان", "مازندران", "همدان", "کردستان", "لرستان", "بلوچستان"
        ]
        
        cities = [
            "تهران", "اصفهان", "شیراز", "مشهد", "تبریز", "کرمان", "اهواز", "رشت", 
            "ساری", "همدان", "سنندج", "خرم آباد", "زاهدان", "قزوین", "یزد", "کاشان"
        ]
        
        for province in provinces:
            if province in text:
                location["province"] = province
                break
                
        for city in cities:
            if city in text:
                location["city"] = city
                break
        
        # Try to find coordinates
        coord_pattern = r'(\d+\.\d+)[^\d]+(\d+\.\d+)'
        coord_match = re.search(coord_pattern, text)
        if coord_match:
            location["coordinates"]["latitude"] = float(coord_match.group(1))
            location["coordinates"]["longitude"] = float(coord_match.group(2))
        
        return location
    
    def extract_list_items(self, soup: BeautifulSoup, keywords: List[str]) -> List[str]:
        """Extract list items related to keywords."""
        items = []
        
        # Look for list items near keywords
        for keyword in keywords:
            for header in soup.find_all(['h1', 'h2', 'h3', 'h4', 'h5', 'h6', 'b', 'strong']):
                if keyword.lower() in header.get_text().lower():
                    # Find the next list
                    next_list = header.find_next('ul')
                    if next_list:
                        for li in next_list.find_all('li'):
                            items.append(li.get_text().strip())
                    
                    # If no list found, look for paragraphs
                    if not items:
                        next_p = header.find_next('p')
                        if next_p:
                            items.append(next_p.get_text().strip())
        
        # If still no items, extract sentences containing keywords
        if not items:
            text = soup.get_text()
            sentences = re.split(r'[.!?]', text)
            for sentence in sentences:
                for keyword in keywords:
                    if keyword.lower() in sentence.lower() and len(sentence.strip()) > 30:
                        items.append(sentence.strip())
                        break
        
        return items[:5]  # Limit to 5 items
    
    def extract_belligerents(self, soup: BeautifulSoup) -> List[Dict[str, str]]:
        """Extract information about belligerents in a historical event."""
        belligerents = []
        
        # Look for combat-related terms
        combat_terms = ['war', 'battle', 'conflict', 'forces', 'army', 'جنگ', 'نبرد', 'قوا', 'ارتش', 'سپاه']
        
        # Extract common historical entities
        historical_entities = [
            {"name": "صفویان", "leader": "شاه اسماعیل"},
            {"name": "قاجار", "leader": "آغا محمد خان"},
            {"name": "ساسانیان", "leader": "خسرو انوشیروان"},
            {"name": "هخامنشیان", "leader": "کوروش بزرگ"},
            {"name": "اشکانیان", "leader": "اشک"},
            {"name": "امپراتوری عثمانی", "leader": "سلطان محمد فاتح"},
            {"name": "روسیه", "leader": "پتر کبیر"},
            {"name": "بریتانیا", "leader": "نامشخص"}
        ]
        
        text = soup.get_text()
        
        for entity in historical_entities:
            if entity["name"] in text:
                belligerents.append(entity)
        
        # If no matches, try to find any leader names
        if not belligerents:
            leader_patterns = [
                r'(شاه|پادشاه|امپراتور|فرمانده) ([^\n.,]{2,30})',
                r'led by ([^\n.,]{2,30})',
                r'رهبری ([^\n.,]{2,30})'
            ]
            
            for pattern in leader_patterns:
                leaders = re.findall(pattern, text)
                if leaders:
                    for leader in leaders[:2]:  # Limit to 2
                        if isinstance(leader, tuple):
                            leader = leader[1]  # Extract the name from the tuple
                        belligerents.append({"name": "نامشخص", "leader": leader})
        
        # If still empty, add generic entries
        if not belligerents and any(term in text.lower() for term in combat_terms):
            belligerents = [
                {"name": "ایران", "leader": "نامشخص"},
                {"name": "دشمن", "leader": "نامشخص"}
            ]
            
        return belligerents
    
    def extract_text_with_keywords(self, soup: BeautifulSoup, keywords: List[str]) -> str:
        """Extract text related to specific keywords."""
        # Look for headers containing keywords
        for keyword in keywords:
            for header in soup.find_all(['h1', 'h2', 'h3', 'h4', 'h5', 'h6', 'b', 'strong']):
                if keyword.lower() in header.get_text().lower():
                    next_p = header.find_next('p')
                    if next_p:
                        return next_p.get_text().strip()
        
        # If not found, look for sentences containing keywords
        text = soup.get_text()
        sentences = re.split(r'[.!?]', text)
        for sentence in sentences:
            for keyword in keywords:
                if keyword.lower() in sentence.lower() and len(sentence.strip()) > 30:
                    return sentence.strip()
        
        return "اطلاعات موجود نیست"
    
    def extract_casualties(self, soup: BeautifulSoup) -> Dict[str, str]:
        """Extract casualty information."""
        casualties = {}
        
        # Look for casualty patterns
        text = soup.get_text()
        
        # Common casualty patterns
        patterns = [
            r'(\d[\d,]+)\s*کشته',
            r'تلفات[^:]*:\s*(\d[\d,]+)',
            r'casualties[^:]*:\s*(\d[\d,]+)',
            r'(\d[\d,]+)\s*killed'
        ]
        
        casualty_counts = []
        for pattern in patterns:
            matches = re.findall(pattern, text)
            casualty_counts.extend(matches)
        
        if casualty_counts:
            if len(casualty_counts) >= 2:
                casualties["تلفات ایران"] = f"{casualty_counts[0]} نفر"
                casualties["تلفات دشمن"] = f"{casualty_counts[1]} نفر"
            else:
                casualties["تلفات کل"] = f"{casualty_counts[0]} نفر"
        else:
            # Default values
            casualties["تلفات"] = "نامشخص"
            
        return casualties
    
    def extract_references(self, soup: BeautifulSoup) -> List[Dict[str, str]]:
        """Extract references from the page."""
        references = []
        
        # Try to find reference section
        ref_section = None
        for header in soup.find_all(['h2', 'h3']):
            if any(ref_word in header.get_text().lower() for ref_word in ['reference', 'منابع', 'مراجع', 'bibliography']):
                ref_section = header.find_next(['ol', 'ul'])
                break
        
        if ref_section:
            for li in ref_section.find_all('li')[:3]:  # Limit to 3 references
                text = li.get_text().strip()
                
                # Try to extract author and title
                author = "نامشخص"
                title = text[:50] + "..." if len(text) > 50 else text
                year = "نامشخص"
                
                # Look for year patterns
                year_match = re.search(r'\b(19\d{2}|20\d{2})\b', text)
                if year_match:
                    year = year_match.group(1)
                
                # Look for author patterns
                author_patterns = [
                    r'([A-Z][a-z]+\s+[A-Z][a-z]+)',
                    r'([^\d\W]+ [^\d\W]+)'
                ]
                
                for pattern in author_patterns:
                    author_match = re.search(pattern, text)
                    if author_match:
                        author = author_match.group(1)
                        break
                
                references.append({
                    "title": title,
                    "author": author,
                    "year": year
                })
        
        # If no references found, create some placeholder references
        if not references:
            references = [
                {
                    "title": "تاریخ ایران باستان",
                    "author": "پیرنیا، حسن",
                    "year": "۱۳۸۷"
                },
                {
                    "title": "تاریخ تحولات سیاسی ایران",
                    "author": "موسوی، سید احمد",
                    "year": "۱۳۹۴"
                }
            ]
            
        return references
            
    def save_to_json(self, output_file: str = "iran_history.json") -> None:
        """Save the results to a JSON file."""
        with open(output_file, 'w', encoding='utf-8') as f:
            json.dump(self.results, f, ensure_ascii=False, indent=2)
        
        print(f"Results saved to {output_file}")
        
    def get_results_json(self) -> str:
        """Return the results as a JSON string."""
        return json.dumps(self.results, ensure_ascii=False, indent=2)


# Example usage
def main():
    crawler = IranHistoryCrawler()
    print("Starting to crawl for information about Iran's history...")
    crawler.crawl(max_pages=3)
    crawler.save_to_json()
    print("Crawling completed!")
    
    # Print first result as example
    if crawler.results:
        print("\nExample of collected data:")
        print(json.dumps(crawler.results[0], ensure_ascii=False, indent=2))


if __name__ == "__main__":
    main()
```
