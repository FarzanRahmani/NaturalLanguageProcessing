```python
import json

# Load JSON data
with open("F:\\Fatemeh\\Education\\Master\\Semester2\\NLP\\Mashahir_rag_3\\Questions\\RetriveByText\\file1_boys.json", "r", encoding="utf-8") as f1:
    file1_data = json.load(f1)

with open("F:\\Fatemeh\\Education\\Master\\Semester2\\NLP\\Mashahir_rag_3\\Questions\\RetriveByText\\file2_girls.json", "r", encoding="utf-8") as f2:
    file2_data = json.load(f2)

merged_data = []

# Convert file1 questions
for idx, entry in enumerate(file1_data, start=1):
    question_text = entry["question"][0]
    options = [opt.split('-', 1)[1].strip() for opt in entry["question"][1:]]
    answer_text = options[entry["answer"] - 1]

    merged_data.append({
        "id": f"file1_{idx}",
        "question": question_text,
        "options": options,
        "answer": answer_text,
        "image": None
    })

# Add file2 questions directly
for idx, entry in enumerate(file2_data, start=1):
    merged_data.append(entry)

# Save merged data
with open("TextOnly_merged_questions.json", "w", encoding="utf-8") as f:
    json.dump(merged_data, f, ensure_ascii=False, indent=2)
```
