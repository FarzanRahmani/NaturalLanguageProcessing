```python
import openai
import os

from dotenv import load_dotenv, find_dotenv
_ = load_dotenv(find_dotenv()) # read local .env file

openai.api_key  = os.getenv('OPENAI_API_KEY')
```

```python
# def get_completion(prompt, model="gpt-3.5-turbo"):
def get_completion(prompt, model="gpt-4.1-nano"):
# def get_completion(prompt, model="gpt-4.1-mini"):
    messages = [{"role": "user", "content": prompt}]
    response = openai.ChatCompletion.create(
        model=model,
        messages=messages,
        temperature=0, # this is the degree of randomness of the model's output
    )
    return response.choices[0].message["content"]
```

```python
# Load zip file
import csv
import sys
import zipfile

# Increase the CSV field size limit
csv.field_size_limit(sys.maxsize)

data = []

# Open the zip file
with zipfile.ZipFile('preprocessed_persian_data.zip', 'r') as z:
    # List all files inside the zip
    file_names = z.namelist()
    print(file_names)  # (optional)

    # Open the first file (assuming it's the CSV)
    with z.open(file_names[0], 'r') as csvfile:
        # Decode bytes to text
        csvfile = (line.decode('utf-8') for line in csvfile)
        reader = csv.DictReader(csvfile)
        for row in reader:
            data.append(row)

# Now 'data' is a list of dictionaries (rows from your CSV)
for entry in data:
    print(entry)
    break
```

```python
structured_data = []

# for entry in data[:100]:
for idx, entry in enumerate(data[:100], start=0):
#     print(idx, entry)
    if idx % 10 == 0:
        print(f"{idx} / {100}")
    QUERY_CONTENT_PLACEHOLDER = entry['content']
    prompt = f"""
        You are a structured data assistant. Your task is to convert a historical text into a JSON structure describing one historical event related to Iran. 

        This event may be one of the following types:
        - War
        - Treaty
        - Revolution

        You must:
        - Detect the type of historical event based on the content I provide.
        - Use the corresponding JSON structure defined below.
        - Complete only **one** JSON structure per request, and **do not include any unused structures**.
        - Output only the completed JSON wrapped between `START_JSON` and `END_JSON`.

        All content (including field values) must be written in **Persian**. 

        If any field is unknown or not mentioned in the input text:
        - Use `""` (empty string) for missing text values.
        - Use `null` for missing numbers.

        The resulting JSON must be fully valid and parsable by `json.loads()` in Python.

        ---

        ### JSON Structure for Historical Wars:

        {WAR_JSON_STRUCTURE_PLACEHOLDER}

        ---

        ### JSON Structure for Historical Treaties:

        {CONTRACTS_JSON_STRUCTURE_PLACEHOLDER}

        ---

        ### JSON Structure for Revolutions:

        {REVOLUTIONS_JSON_STRUCTURE_PLACEHOLDER}

        ---


        ### Additional Rules:

        - Include **all fields** from the selected structure, even if empty.
        - Do **not** alter the JSON schema (no adding/removing fields).
        - Do **not** include any comments inside the final JSON output.
        - Be as detailed and complete as possible in every field.
        - Do not summarize or skip explanation fields. Write full sentences where expected.
        - Your output **must** begin with the label `START_JSON` and end with `END_JSON`.
        - Output **nothing else** outside those two markers.
        - If the event content is ambiguous or not clearly one of the types, choose the **closest** category and proceed.

        ---

        ### Examples of Correct Output:

        Example output for a **war**:
        {SAMPLE_JSON_OUTPUT_WAR_PLACEHOLDER}

        Example output for a **treaty**:
        {SAMPLE_JSON_OUTPUT_CONTRACTS_PLACEHOLDER}

        Example output for a **revolution**:
        {SAMPLE_JSON_OUTPUT_REVOLUTIONS_PLACEHOLDER}


        ---

        Now, please analyze the following text and convert it to the appropriate JSON format:

        {QUERY_CONTENT_PLACEHOLDER}

        """
    response = get_completion(prompt)
    # Extract content between START_JSON and END_JSON
    try:
        extracted_text = response.split('START_JSON')[1].split('END_JSON')[0].strip()
        extracted_json = json.loads(extracted_text)
    except Exception as e:
        print(f"Error parsing JSON for entry: {e}")
        extracted_json = None  # or you could skip this entry if you prefer
    
    
    new_entry = entry.copy()  # So we don't overwrite original 'data'
    new_entry['extracted_content'] = extracted_json
#     new_entry['extracted_content'] = response
#     print(extracted_json)
    structured_data.append(new_entry)
```
