```python
def query(prompt: str, model_idx=0, temperature = 0.0, statistics=True):
    models = [
        "llama3",
        "phi4-mini",
        "qwen2.5:7b",
        "deepseek-r1:14b",
        "gemma3:4b",
        "falcon:7b",
        "ALIENTELLIGENCE/contentsummarizer:latest",
        "hemanth/booksummarizer:latest",
        "moetez/text_summarizer:latest"
    ]
    start = time.time()
    response = requests.post(
        "http://localhost:11434/api/generate",
        json={
            "model": models[model_idx],
            "prompt": prompt,
            "stream": False,
            "temperature": temperature,
            # "response_format": "json",
            # "system": system,
            # "template": template,
            # "raw": False,
            # "keep_alive": "",
        }
    )
    end = time.time()
    response_time = end - start
    if statistics:
        print("MODEL:       {}".format(models[model_idx]))
        print("TEMPERATURE: {}".format(temperature))
        print("TIME:        {:.2f} seconds".format(response_time))
    return response.json()["response"]
```

```python
def read_json_file(file_path):
    try:
        with open(file_path, 'r') as file:
            data = json.load(file)
        return data
    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")
        return None
    except json.JSONDecodeError:
        print(f"Error: Invalid JSON format in '{file_path}'.")
        return None


def read_txt_file(file_path):
    try:
        with open(file_path, 'r') as file:
            content = file.read()
        return content
    except FileNotFoundError:
        print(f"Error: File '{file_path}' not found.")
        return None
```

# Second Approach: Try to Fill JSON Entries Iteratively by Search and SLMs

```python
war_names = [
    "جنگ ایران و عراق",
    "جنگ شاهنشاهی ساسانی و امپراتوری بیزانس",
    "حمله اعراب به ایران",
]


search_queries = [
    "تاریخ دقیق شروع و پایان",
    "استان ها و شهر های درگیر در",
    "لیست دلایل وقوع",
    "نام کشور های درگیر در",
    "نام رهبران درگیر در",
    "نتیجه نهایی",
    "تلفات نظامیان طرفین در",
    "تلفات غیر نظامیان طرفین در",
    "تاثیر سیاسی",
    "تاثیر اقتصادی",
    "تاثیر اجتماعی",
    "تاثیر فرهنگی",
    "افراد و نقش هر یک در",
    "منابع، سال انتشار منبع و نویسنده منبع درباره",
]


base_prompt = "فقط query را به فارسی از متن مقابل استخراج کن و برگردان و چیز بیشتری بیان نکن"
save_dir = "/content/drive/MyDrive/HW#1/outputs_files"
```

```python
def convert_persian_digits_to_english(text):
    persian_numbers = '۰۱۲۳۴۵۶۷۸۹٫'
    english_numbers = '0123456789.'
    translation_table = str.maketrans(persian_numbers, english_numbers)
    return text.translate(translation_table)


def extract_json_from_text(text):
    pattern = r'\{((?:[^{}]|\{(?:[^{}]|\{[^{}]*\})*\})*)\}'
    matches = re.finditer(pattern, text, re.DOTALL)
    extracted = [match.group(0) for match in matches]
    json_str = json5.loads(extracted[0])
    return json_str


def save_json_to_file(json_data, output_file):
    with open(output_file, "w", encoding="utf-8") as f:
        json.dump(json_data, f, ensure_ascii=False, indent=4)
```

```python
search = DuckDuckGoSearchRun()

for war_name in war_names:
    start = time.time()

    extracted_info = "عنوان جنگ " + war_name + "\n\n\n\n\n\n"

    for idx, base_search_query in tqdm(enumerate(search_queries)):
        search_query = base_search_query + " " +  war_name
        prompt = base_prompt.replace("query", search_query)

        web_content = search.run(search_query)
        time.sleep(0.2)

        summary = query(prompt + ": " + web_content, model_idx=0, statistics=False)
        summary = convert_persian_digits_to_english(summary)

        extracted_info += search_query + ": " + summary + "\n\n\n\n\n\n"

    final_prompt = PROMPT.replace("QUERY_CONTENT_PLACEHOLDER", extracted_info)

    json_content = query(final_prompt, model_idx=0, statistics=False)

    save_path = os.path.join(save_dir, war_name + ".json")
    json_data = extract_json_from_text(json_content)
    save_json_to_file(json_data, save_path)

    end = time.time()
    retrive_time = end - start


    print("RETRIVE TIME:         {:.3f} seconds".format(retrive_time))
    print("")
```
